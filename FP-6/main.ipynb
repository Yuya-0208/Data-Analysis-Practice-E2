## Classification Analysis of Wholesale Customers

## 1. Introduction
The objective of this analysis is to predict the sales channel of customers based on their annual spending across various product categories. The dataset used is the **Wholesale Customers Data**, which contains spending information on categories such as Fresh, Milk, Grocery, Frozen, Detergents_Paper, and Delicassen.

The target variable is **Channel**, which indicates the type of customer:
* **Channel 1**: Horeca (Hotel/Restaurant/Cafe)
* **Channel 2**: Retail

By analyzing these spending patterns, we aim to build a machine learning model that can accurately classify new customers.

## 2. Methodology
To perform the classification task, I utilized the **K-Nearest Neighbors (KNN)** algorithm. The analysis process followed these steps:

1.  **Preprocessing**: Since KNN is a distance-based algorithm, features with larger scales can disproportionately influence the result. To prevent this, I applied **StandardScaler** to normalize all feature variables (mean=0, variance=1).
2.  **Data Splitting**: The dataset was split into a **training set (67%)** and a **testing set (33%)** to evaluate the model's performance on unseen data.
3.  **Validation Strategy**: To avoid overfitting and select the optimal model complexity, I performed a validation loop. I tested the number of neighbors (k) ranging from 1 to 30 and compared the training and testing accuracy scores.

## 3. Results

### Result 1: Model Optimization (Validation)
I evaluated the KNN model's performance by varying the hyperparameter k (number of neighbors). The figure below illustrates the relationship between k and the model's accuracy.

* **Observation**: When k is very small (e.g., k=1), the training accuracy is perfect, but the test accuracy is lower, indicating **overfitting**. As k increases, the model generalizes better. However, if k becomes too large, the accuracy starts to drop due to underfitting.
* **Decision**: Based on the validation curve, I selected the k value that maximized the accuracy on the test set.

From this analysis, it can be said that the best k is 7.

### Result 2: Final Model Performance
Using the optimal parameter (**k = 7**), I evaluated the final model on the test set. Figure 2 shows the **Confusion Matrix**, which breaks down the predictions into correct classifications and errors for each channel.

This visualization helps us understand not just the overall accuracy, but also how well the model distinguishes between 'Horeca' and 'Retail' customers.

## 4. Conclusion

These analysis confirms that customer spending patterns—specifically across categories like Fresh goods, Grocery, and Detergents—are highly effective indicators of their business type (Horeca vs. Retail). Also, these analysis enable us to develop an effective marketing strategy.

**Key Takeaways:**
* **High Reliability:** The developed model achieved an accuracy of approximately **90%**. This reliability allows for automated classification of new customers without manual verification.
* **Actionable Insights:** By correctly identifying the customer channel, the business can optimize its supply chain:
    * **For Retail clients:** Focus on stocking Grocery and Detergents/Paper products.
    * **For Horeca clients:** Ensure high availability of Fresh foods and Frozen goods.
