# parse_data.ipynb

This notebook parses the data files used for the FP-2 assignment. 

<br>
<br>

First let's read the attached data file:

```python
import pandas as pd

df0 = pd.read_csv('Wholesale customers data.csv')

df0.describe()
```

<br>
<br>

The dependent and independent variables variables (DVs and IVs) that we are interested in are:

**DVs**:
- Delicassen  (Delicatessen, how much money customers spend on it.)

**IVs**:
- Fresh ("Fresh" column in the CSV file)
- Milk ("Milk" column in the CSV file)
- Grocery ("Grocery" column in the CSV file)
- Frozen ("Frozen" column in the CSV file)

<br>
<br>

Let's extract the relevant columns:

```python

df = df0[  ['Fresh', 'Milk', 'Grocery', 'Frozen'] ]
df.describe()
```

<br>
<br>

Next let's use the `rename` function to give the columns simpler variable names:

```python
df = df.rename( columns={'Fresh':'Fre', 'Grocery':'Gro', 'Frozen':'Fro'} )

df.describe()
```
