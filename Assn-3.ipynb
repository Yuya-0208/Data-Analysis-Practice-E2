## Figure1
Use `data-fig1.npy` for this figure.

```python
%matplotlib inline

import numpy as np
from matplotlib import pyplot as plt

a   = np.load( 'data-fig1.npy' )

plt.figure()
plt.plot(a, color='black', marker='o')
plt.xlabel('Time')
plt.ylabel('Position')
plt.show()
```

## Figure 2
Use `data-fig2.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt

b   = np.load( 'data-fig2.npy' )

plt.figure()
plt.plot(b[:,0], color='red', label='Red')
plt.plot(b[:,1], color='green', label='Green')
plt.plot(b[:,2], color='blue', label='Blue')


# 軸ラベルと凡例の設定
plt.xlabel('X value')
plt.ylabel('Y value')
plt.legend() 
plt.show()
```

## Figure3
Use `data-fig3.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt

data = np.load('data-fig3.npy') 
x_data = data[:, 0]
y_data = data[:, 1]

plt.figure(figsize=(7, 7))
plt.scatter(
    x_data, y_data, 
    color='blue',            
    alpha=0.4,              
    s=200,                   
    edgecolors='none'        
)

plt.xlabel('X value')
plt.ylabel('Y value')

plt.show()
```

## Figure 4
Use `data-fig4.npy` for this figure.

```python
import numpy as np
from matplotlib import pyplot as plt

values = np.load('data-fig4.npy') 

categories = ['A', 'B', 'C', 'D', 'E']

plt.figure(figsize=(7, 5))

plt.bar(
    categories, 
    values, 
    color='gray' 
)

plt.xlabel('Category')
plt.ylabel('Value')

plt.show()
```

## Figure 5
Use data-fig5.npy for this figure.

Hint: Use 2D visualization and the "RdBu" colormap.

```python
data = np.load('data-fig5.npy') 

plt.figure(figsize=(7, 5))

im = plt.imshow(
    data, 
    cmap='RdBu',           
    origin='lower',        
    aspect='auto'    
)

plt.colorbar(im) 

plt.xlabel('X value') 
plt.ylabel('Y value')

plt.show()
```
